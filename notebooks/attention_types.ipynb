


### 14. Performer-Style Attention
Performer-style attention uses random feature methods to approximate softmax attention while avoiding explicit construction of the full $L\times L$ matrix.

> Use cases
- Long sequences
- Memory-constrained training
- Efficient Transformer research
- Large-scale sequence modeling

> Time complexity:With feature dimension $r$:

$O(Ldr)$

Space complexity is approximately:

$O(Lr)$

If $r$ is fixed or grows slowly, this is effectively linear in sequence length. The approximation introduces a quality–efficiency trade-off.

---

### 15. Memory-Augmented Attention
**Memory-augmented attention** lets a model attend not only to the current sequence but also to an external or persistent memory bank.

The memory may contain:

- Previous hidden states
- Retrieved documents
- Learned memory slots
- External knowledge
- Cached representations

> Example: A question-answering system can use the query as a query vector and attend to retrieved passages stored in an external memory bank.

> Use cases
- Retrieval-augmented generation
- Long-term dialogue memory
- Knowledge-grounded question answering
- Neural computers
- Recommendation systems
- Episodic memory in agents

> Time complexity:If there are $L$ sequence queries and $M$ memory entries:

$O(LMd)$

If retrieval first narrows the memory to $k$ entries, the attention cost becomes:

$O(Lkd), \quad k\ll M$

The total system cost also includes the cost of searching or retrieving the memory.

---

### 16. Grouped-Query Attention
In **grouped-query attention**, multiple query heads share the same key and value heads. It lies between multi-head attention and multi-query attention.

Suppose there are $h_q$ query heads and $h_{kv}$ key/value heads, where:

$h_{kv}<h_q$

Queries retain multiple distinct heads, but keys and values are stored in fewer groups.

> Example: A model may use 32 query heads but only 8 key/value heads. Every group of four query heads shares one key/value head.

> Use cases
- Efficient large-language-model inference
- Reducing key-value cache memory
- Decoder-only Transformers
- Long-context generation

> Time complexity:The theoretical dense attention computation remains approximately:

$O(L^2d)$

However, key/value projection and cache storage are reduced. During autoregressive inference, the key-value cache is reduced by approximately the factor:

$\frac{h_q}{h_{kv}}$

compared with ordinary multi-head attention, assuming equal head dimensions.

---

### 18. FlashAttention
**FlashAttention** is an exact attention algorithm that improves speed and memory efficiency through tiling and recomputation. It does not fundamentally change the mathematical attention function or its quadratic asymptotic time complexity.

Instead of materializing the entire attention matrix in high-bandwidth memory, it processes blocks of queries, keys, and values while maintaining numerically stable softmax statistics.

> Use cases
- Faster Transformer training
- Longer sequences within the same GPU memory
- Large language models
- Vision and multimodal Transformers

> Complexity

Mathematically:

$O(L^2d)$

time complexity remains. However, memory usage for the attention intermediate can be reduced from approximately:

$O(L^2)$

to roughly:

$O(Ld)$

for the main working memory, depending on implementation details. FlashAttention is therefore an **implementation optimization**, not a sparse or linear-attention approximation.

---

### 19. Channel or Feature Attention
Not all attention operates across tokens. **Channel attention** assigns importance to feature channels rather than positions in a sequence or spatial locations in an image.

For example, a convolutional network may learn that color, texture, or edge-related channels are more important for a particular image.

> Example: In an image-recognition model, channel attention may amplify channels representing fur texture while reducing irrelevant channels for recognizing a dog.

> Use cases
- Squeeze-and-Excitation networks
- Convolutional computer vision
- Image classification
- Object detection
- Feature recalibration

> Time complexity:If there are $C$ channels and the attention mechanism uses a reduced dimension $r
after compression, the cost may be approximately:

$O(Cr)$

or $O(C^2)$ for a dense channel-interaction matrix. This is separate from token self-attention complexity.

---

### 20. Spatial Attention
**Spatial attention** assigns importance to different spatial positions in an image or feature map. It answers the question:

> Which regions of the input are important?

> Example: For an image containing a person and a background, spatial attention may assign high weights to the person and lower weights to the sky or ground.

> Use cases
- Object detection
- Image segmentation
- Image captioning
- Medical-image analysis
- Convolutional attention modules

> Time complexity:For $N=H\times W$ spatial positions and dense global attention:

$O(N^2d)$

For local spatial attention with window size $w$:

$O(Nwd)$

This is why local-window methods are useful for high-resolution images.

---

### 21. Temporal Attention
**Temporal attention** assigns different weights to time steps in sequential data. It enables a model to focus on the most informative moments.

> Example: In video classification, the model may attend strongly to frames showing a person falling while ignoring frames showing a static background.

> Use cases
- Video understanding
- Speech recognition
- Human-activity recognition
- Time-series forecasting
- Sensor-data analysis
- Healthcare monitoring

> Time complexity:For $T$ time steps and feature dimension $d$, dense temporal attention has:

$O(T^2d)$

Local temporal attention with window size $w$ has:

$O(Twd)$

---

### 22. Hierarchical Attention
**Hierarchical attention** applies attention at multiple levels of a structure. A document, for example, can be represented using:

1. Word-level attention within each sentence
2. Sentence-level attention across the document

The model first identifies important words and then identifies important sentences.

> Example: For sentiment classification, word-level attention may highlight “excellent” and “disappointing,” while sentence-level attention may identify the review sentence that most strongly expresses the overall opinion.

> Use cases
- Long-document classification
- News summarization
- Legal-document analysis
- Scientific-paper understanding
- Meeting and transcript summarization

> Time complexity:Suppose a document has $S$ sentences, each containing approximately $W$ words. Word-level attention costs roughly:

$O(SW^2d)$

Sentence-level attention costs:

$O(S^2d)$

Total complexity is approximately:

$O(SW^2d+S^2d)$

This can be substantially cheaper than attending over all $SW$ words at once:

$O((SW)^2d)$

but it imposes a hierarchical structure that may lose some direct word-to-word interactions across distant sentences.

---

### 23. Hard Attention and Soft Attention

### Soft attention

**Soft attention** assigns a continuous probability distribution over inputs. Several or all inputs can receive nonzero weights.

Example:

$\alpha=[0.1,0.7,0.2]$

The output is a differentiable weighted sum.

> Complexity

If all $m$ items are considered for each of $n$ queries:

$O(nmd)$

#> Use cases
- Standard Transformers
- Differentiable sequence-to-sequence systems
- Image captioning
- Text classification

## Hard attention

**Hard attention** selects one item or a small subset, often using sampling, top-k selection, or discrete decisions.

Example:

$[0,1,0,0]$

Only one input is selected.

> Complexity

If selecting $k$ items per query after efficient search, aggregation may cost approximately:

$O(nkd), \quad k\ll m$

But finding the top items may itself cost $O(nm)$, unless an index or approximate-search structure is used.

#> Use cases
- Memory retrieval
- Hard routing
- Sparse expert systems
- Pointer networks
- Computationally constrained models

The key difference is that soft attention is easy to train with gradient descent, whereas hard attention may require reinforcement learning, straight-through estimators, or other approximations.

## 24. Self-Attention Versus Cross-Attention: Key Difference

| Feature | Self-attention | Cross-attention |
|---|---|---|
| Queries come from | Same sequence | One sequence or modality |
| Keys and values come from | Same sequence | Another sequence or modality |
| Main purpose | Build contextual representations | Transfer or retrieve information across representations |
| Example | Words attending to other words | Decoder attending to encoder output |
| Complexity | $O(L^2d)$ | $O(nmd)$ |

---

## 25. Complexity Summary

Let $L$ be sequence length, $d$ hidden dimension, $w$ local window size, $M$ memory size, and $h$ number of heads.

| Attention type | Typical time complexity | Typical attention memory | Main advantage |
|---|---:|---:|---|
| Dense self-attention | $O(L^2d)$ | $O(L^2)$ | Full global interaction |
| Cross-attention | $O(nmd)$ | $O(nm)$ | Connects different sequences or modalities |
| Additive attention | $O(nm(a+d_v))$ | $O(nm)$ | Flexible learned scoring |
| Multiplicative attention | $O(nmd)$ | $O(nm)$ | Fast similarity computation |
| Multi-head attention | $O(L^2d)$ | $O(L^2)$ | Multiple representation subspaces |
| Local attention | $O(Lwd)$ | $O(Lw)$ | Efficient local context |
| Sparse attention | $O(Lsd)$ | $O(Ls)$ | Flexible reduced connectivity |
| Global-plus-local attention | $O(L(w+g)d)$ approximately | Reduced versus dense | Combines local and global context |
| Linear attention | Often $O(Ld^2)$ or $O(Ldr)$ | Approximately linear | Scales to long sequences |
| FlashAttention | $O(L^2d)$ | Approximately $O(Ld)$ working memory | Exact but memory-efficient computation |
| Multi-query attention | Approximately dense attention cost | Much smaller KV cache | Faster decoding |
| Grouped-query attention | Approximately dense attention cost | Reduced KV cache | Balance of quality and efficiency |
| Hierarchical attention | Problem-dependent | Lower than full document attention in many cases | Uses document structure |

## 26. Which Attention Type Should You Use?

| Problem | Suitable attention mechanism |
|---|---|
| Ordinary text classification | Bidirectional self-attention |
| Autoregressive text generation | Causal self-attention |
| Machine translation | Encoder self-attention plus decoder cross-attention |
| Text-to-image generation | Cross-attention between image and text representations |
| Very long documents | Local, sparse, hierarchical, or linear attention |
| Fast LLM inference | Multi-query or grouped-query attention with KV caching |
| High-resolution images | Windowed or hierarchical spatial attention |
| Video understanding | Temporal, spatial, local, or factorized attention |
| External knowledge retrieval | Memory-augmented or cross-attention |
| Maximum accuracy on moderate sequence lengths | Dense multi-head attention |
| GPU memory bottleneck | FlashAttention or checkpointing |

## Final takeaway

The most important distinction is between **what attends to what** and **how the similarity is computed**. Self-attention operates within one sequence, cross-attention connects different representations, additive and multiplicative attention describe scoring functions, and local, sparse, linear, multi-query, and FlashAttention methods primarily address efficiency. Standard dense multi-head self-attention remains powerful because it provides unrestricted global interaction, but its $O(L^2d)$ time and $O(L^2)$ score-memory costs motivate efficient alternatives for long inputs.
